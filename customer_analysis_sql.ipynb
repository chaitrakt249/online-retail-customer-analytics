```python
result = conn.execute("""
WITH customer_summary AS (
    SELECT
        "Customer ID",
        COUNT(DISTINCT Invoice) AS Orders,
        SUM(Revenue) AS Revenue
    FROM online_retail
    GROUP BY "Customer ID"
)

SELECT
    CASE
        WHEN Orders > 1 THEN 'Repeat Customer'
        ELSE 'One-Time Customer'
    END AS Customer_Type,
    COUNT(*) AS Customers,
    ROUND(SUM(Revenue), 2) AS Total_Revenue,
    ROUND(
        SUM(Revenue) * 100.0 /
        SUM(SUM(Revenue)) OVER (),
        2
    ) AS Revenue_Share_Percent
FROM customer_summary
GROUP BY Customer_Type
ORDER BY Total_Revenue DESC
""").df()

display(result)
```

```python
result = conn.execute("""
WITH product_country_sales AS (
    SELECT
        Country,
        Description,
        ROUND(SUM(Revenue), 2) AS Revenue
    FROM online_retail
    GROUP BY Country, Description
),

ranked_products AS (
    SELECT
        Country,
        Description,
        Revenue,
        RANK() OVER (
            PARTITION BY Country
            ORDER BY Revenue DESC
        ) AS Product_Rank
    FROM product_country_sales
)

SELECT
    Country,
    Description,
    Revenue,
    Product_Rank
FROM ranked_products
WHERE Product_Rank <= 3
ORDER BY Country, Product_Rank
""").df()

display(result)
```

```python
result = conn.execute("""
WITH monthly_sales AS (
    SELECT
        DATE_TRUNC('month', InvoiceDate) AS Month,
        SUM(Revenue) AS Revenue
    FROM online_retail
    GROUP BY 1
),

previous_month AS (
    SELECT
        Month,
        Revenue,
        LAG(Revenue) OVER (
            ORDER BY Month
        ) AS Previous_Month_Revenue
    FROM monthly_sales
)

SELECT
    Month,
    ROUND(Revenue, 2) AS Revenue,
    ROUND(Previous_Month_Revenue, 2) AS Previous_Month_Revenue,
    ROUND(
        (Revenue - Previous_Month_Revenue)
        * 100.0
        / Previous_Month_Revenue,
        2
    ) AS Revenue_Growth_Percent
FROM previous_month
ORDER BY Month
""").df()

display(result)
```

```python
result = conn.execute("""
WITH customer_revenue AS (
    SELECT
        "Customer ID",
        SUM(Revenue) AS Revenue
    FROM online_retail
    GROUP BY "Customer ID"
),

ranked AS (
    SELECT
        "Customer ID",
        Revenue,
        RANK() OVER (
            ORDER BY Revenue DESC
        ) AS Revenue_Rank
    FROM customer_revenue
),

totals AS (
    SELECT
        SUM(Revenue) AS Total_Revenue
    FROM customer_revenue
)

SELECT
    r."Customer ID",
    ROUND(r.Revenue, 2) AS Revenue,
    r.Revenue_Rank,
    ROUND(
        r.Revenue * 100.0 / t.Total_Revenue,
        2
    ) AS Revenue_Share_Percent
FROM ranked r
CROSS JOIN totals t
ORDER BY r.Revenue_Rank
LIMIT 20
""").df()

display(result)
```

```python
# Recreate RFM analysis in Colab

import pandas as pd

# Load transaction data from DuckDB
cleaned_df = conn.execute("""
    SELECT *
    FROM online_retail
""").df()

# Make sure InvoiceDate is a datetime
cleaned_df["InvoiceDate"] = pd.to_datetime(cleaned_df["InvoiceDate"])

# Analysis date
analysis_date = cleaned_df["InvoiceDate"].max() + pd.Timedelta(days=1)

# Calculate RFM
rfm = (
    cleaned_df.groupby("Customer ID")
    .agg(
        Recency=("InvoiceDate", lambda x:
                 (analysis_date - x.max()).days),
        Frequency=("Invoice", "nunique"),
        Monetary=("Revenue", "sum")
    )
    .reset_index()
)

# RFM scores
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

# Customer segmentation
def segment_customer(row):

    R = row["R_Score"]
    F = row["F_Score"]
    M = row["M_Score"]

    if R >= 4 and F >= 4 and M >= 4:
        return "Champions"

    elif R >= 3 and F >= 4 and M >= 3:
        return "Loyal Customers"

    elif R >= 3 and M >= 4:
        return "Big Spenders"

    elif R >= 4 and F >= 2:
        return "Potential Loyalists"

    elif R >= 4 and F <= 2:
        return "New Customers"

    elif R <= 2 and F >= 3 and M >= 3:
        return "At Risk"

    elif R <= 2 and F >= 4:
        return "Can't Lose Them"

    elif R <= 2 and F <= 2 and M <= 2:
        return "Lost Customers"

    else:
        return "Regular Customers"

rfm["Customer_Segment"] = rfm.apply(
    segment_customer,
    axis=1
)

print("RFM dataset created successfully!")
print("Customers:", len(rfm))

display(rfm.head())
```

```python
rfm.to_csv(
    "rfm_customer_segments.csv",
    index=False
)

print("rfm_customer_segments.csv created successfully!")
```
